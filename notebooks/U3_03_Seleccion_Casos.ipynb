# 🧩 Cuaderno 03 — Selección de Equipos y Caso de Estudio

**Métodos Numéricos · Unidad 3: Sistemas de Ecuaciones Lineales**

Este cuaderno lo usa **el jefe de equipo** (una sola vez): registra al equipo y
**asigna de forma determinista** el **Caso de Estudio** (T1–T17) a partir del
**Número de Control (NC)** de los integrantes.

> ⚠️ **Aquí no se resuelve el caso.** Cuando el equipo quede registrado, pasen al
> **Cuaderno 04**, donde se hace el análisis.

### 🎯 Cómo funciona la asignación
1. El **tamaño del equipo** (2–6 integrantes) define el **bloque** de casos (A–E).
2. Una **semilla** calculada con los dígitos de los NC elige **cuál** caso del bloque les toca.
3. El resultado es **reproducible**: los mismos NC siempre dan el mismo caso.

```python
#@title ⚙️ Entorno y conexión al backend (ejecuta PRIMERO)
import os, json, urllib.request
try:
    import google.colab
    ES_COLAB = True
except ImportError:
    ES_COLAB = False

ARCHIVO = "robomat.py"
URL = "https://raw.githubusercontent.com/ITH-MGL-MN/U3_Sistemas_de_ecuaciones_lineales/main/lib/robomat.py"


def _buscar_robomat(nombre=ARCHIVO, saltos=6):
    """Busca robomat.py junto al notebook, en la carpeta actual y en lib/ de los padres."""
    bases = [os.getcwd()]
    ruta_nb = globals().get("__vsc_ipynb_file__")      # VS Code
    if ruta_nb:
        bases.append(os.path.dirname(os.path.abspath(ruta_nb)))
    for base in list(bases):
        b = base
        for _ in range(saltos):
            for cand in (os.path.join(b, nombre), os.path.join(b, "lib", nombre)):
                if os.path.exists(cand):
                    return os.path.abspath(cand)
            padre = os.path.dirname(b)
            if padre == b:
                break
            b = padre
    return None


ruta_robomat = _buscar_robomat()
if ruta_robomat is None:
    print("robomat.py no está en disco; descargando del repositorio...")
    urllib.request.urlretrieve(URL, ARCHIVO)
    ruta_robomat = os.path.abspath(ARCHIVO)

exec(open(ruta_robomat, encoding="utf-8").read(), globals())
print(f"✅ Entorno cargado desde: {ruta_robomat}")

# ===== Conexión al backend (Apps Script) =====
# El profesor llena estos dos valores tras desplegar App Script/proyectos_u3.gs
APPS_SCRIPT_URL = ""   #@param {type:"string"}
WEBHOOK_TOKEN   = ""   #@param {type:"string"}

BACKEND_LISTO = bool(APPS_SCRIPT_URL.strip() and WEBHOOK_TOKEN.strip())

def _post(accion, **campos):
    """POST al Apps Script. Devuelve el dict 'data' o lanza excepción."""
    import urllib.request
    payload = {"accion": accion, "token": WEBHOOK_TOKEN, **campos}
    req = urllib.request.Request(
        APPS_SCRIPT_URL,
        data=json.dumps(payload).encode("utf-8"),
        headers={"Content-Type": "application/json"})
    with urllib.request.urlopen(req, timeout=30) as r:
        resp = json.loads(r.read().decode("utf-8"))
    if resp.get("status") != "ok":
        raise RuntimeError(resp.get("message", "Error del backend"))
    return resp.get("data", {})

print("✅ Entorno:", "Google Colab" if ES_COLAB else "local")
print("🔌 Backend:", "CONECTADO" if BACKEND_LISTO else "sin configurar (modo local)")
```

    "T15": ("Cercha Espacial 3D para Brazo/Torre de Soporte", "D", 5),

    "T17": ("Flujo de Carga Linealizado (DC Power Flow) en Red de Potencia", "E", 6),
}
BLOQUE_POR_TAM = {2:"A", 3:"B", 4:"C", 5:"D", 6:"E"}
CASOS_POR_BLOQUE = {
    "A": ["T1","T2","T3","T4"], "B": ["T5","T6","T7","T8","T9"],
    "C": ["T10","T11","T12","T13","T14"], "D": ["T15","T16"], "E": ["T17"],
}

def asignar_caso(ncs):
    n = len(ncs)
    if n not in BLOQUE_POR_TAM:
        raise ValueError("El equipo debe tener entre 2 y 6 integrantes")
    bloque = BLOQUE_POR_TAM[n]
    casos = CASOS_POR_BLOQUE[bloque]
    semilla = sum(int(d) for nc in ncs for d in str(nc) if d.isdigit())
    return bloque, casos[semilla % len(casos)], semilla

print("✅ Catálogo local:", len(CASOS), "casos")
