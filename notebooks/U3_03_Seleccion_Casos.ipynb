### 🎯 Cómo funciona la asignación
1. El **tamaño del equipo** (2–6 integrantes) define el **bloque** de casos (A–E).
2. Una **semilla** calculada con los dígitos de los NC elige **cuál** caso del bloque les toca.
3. El resultado es **reproducible**: los mismos NC siempre dan el mismo caso.


```python
#@title ⚙️ Cargar entorno (ejecuta PRIMERO)
import os, urllib.request
try:
    import google.colab
    ES_COLAB = True
except ImportError:
    ES_COLAB = False

ARCHIVO = "robomat.py"
if ES_COLAB and not os.path.exists(ARCHIVO):
    URL = "https://raw.githubusercontent.com/<USUARIO>/<REPO>/main/lib/robomat.py"
    urllib.request.urlretrieve(URL, ARCHIVO)

exec(open(ARCHIVO, encoding="utf-8").read(), globals())
print("✅ Entorno cargado:", "Google Colab" if ES_COLAB else "local")
```

```python
#@title 📚 Catálogo de casos (metadatos — no modificar)
CASOS = {
    "T1":  ("Análisis de Corrientes de Malla en Circuito de CD", "Bloque A · 2 integrantes"),
    "T2":  ("Análisis de Nodos de Voltaje con Fuentes de Corriente", "Bloque A · 2 integrantes"),
    "T3":  ("Red de Filtros Pasivos RC en Escalera de 3 Etapas", "Bloque A · 2 integrantes"),
    "T4":  ("Conducción Térmica 1D en Barra Discretizada", "Bloque A · 2 integrantes"),
    "T5":  ("Cercha Plana 2D de Brazo Robótico (4 Nodos / 5 Barras)", "Bloque B · 3 integrantes"),
    "T6":  ("Sistema Masa-Resorte de 4 Masas Acopladas", "Bloque B · 3 integrantes"),
    "T7":  ("Deflexión de Viga Continua sobre Múltiples Apoyos", "Bloque B · 3 integrantes"),
    "T8":  ("Repartición de Cargas en Suspensión Neumática Multieje", "Bloque B · 3 integrantes"),
    "T9":  ("Balance de Materia en 4 Reactores CSTR Acoplados", "Bloque B · 3 integrantes"),
    "T10": ("Conducción Térmica 2D en Placa Rectangular", "Bloque C · 4 integrantes"),
    "T11": ("Distribución de Caudales en Red de Tuberías Linealizada", "Bloque C · 4 integrantes"),
    "T12": ("Estequiometría de Combustión Compleja", "Bloque C · 4 integrantes"),
    "T13": ("Modelo Insumo-Producto de Leontief", "Bloque C · 4 integrantes"),
    "T14": ("Separación por Etapas en Columna de Destilación", "Bloque C · 4 integrantes"),
    "T15": ("Cercha Espacial 3D para Brazo/Torre de Soporte", "Bloque D · 5 integrantes"),
    "T16": ("Red de Intercambiadores de Calor Serie/Paralelo", "Bloque D · 5 integrantes"),
    "T17": ("Flujo de Carga Linealizado (DC Power Flow) en Red de Potencia", "Bloque E · 6 integrantes"),
}

BLOQUE_POR_TAM = {2:"A", 3:"B", 4:"C", 5:"D", 6:"E"}
CASOS_POR_BLOQUE = {
    "A": ["T1","T2","T3","T4"],
    "B": ["T5","T6","T7","T8","T9"],
    "C": ["T10","T11","T12","T13","T14"],
    "D": ["T15","T16"],
    "E": ["T17"],
}

def asignar_caso(ncs):
    n = len(ncs)
    bloque = BLOQUE_POR_TAM[n]
    casos = CASOS_POR_BLOQUE[bloque]
    semilla = sum(int(d) for nc in ncs for d in str(nc) if d.isdigit())
    caso = casos[semilla % len(casos)]
    return bloque, caso, semilla

print("✅ Catálogo cargado:", len(CASOS), "casos")
```

## 1. Registra a tu equipo

Escribe el **número de control** de **cada integrante**, separados por comas
(el primero es el jefe). El número de integrantes define el bloque de casos.


```python
#@title 🧑‍🤝‍🧑 Integrantes (NC separados por comas)
ncs_csv = "21340001,21340002,21340003"  #@param {type:"string"}
#@markdown El primer NC es el jefe de equipo. Borra el ejemplo y escribe los reales.

NCS = [nc.strip() for nc in ncs_csv.split(",") if nc.strip()]
print("Integrantes registrados:", len(NCS))
for i, nc in enumerate(NCS, 1):
    print(f"  {i}. NC {nc}" + ("  (jefe)" if i == 1 else ""))
```

```python
#@title 🎲 Asignar caso de estudio
if not (2 <= len(NCS) <= 6):
    print("⚠️ El equipo debe tener entre 2 y 6 integrantes.")
else:
    bloque, caso, semilla = asignar_caso(NCS)
    titulo, detalle = CASOS[caso]
    print("=" * 60)
    print(f"BLOQUE {bloque}  ·  {len(NCS)} integrantes")
    print(f"CASO ASIGNADO: {caso}")
    print(f"📌 {titulo}")
    print(f"   {detalle}")
    print(f"Semilla (suma de dígitos de los NC): {semilla}")
    print("=" * 60)
    print()
    print("📖 Ficha técnica en el repositorio:")
    print(f"   docs/casos/caso_{caso.lower()}.md")
```
